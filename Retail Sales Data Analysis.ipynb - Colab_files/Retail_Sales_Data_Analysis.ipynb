```python
# Connect the returns and people sheet #
def clean_key(series):
  return(
      series.astype("string")
      .str.replace("\u00a0"," ",regex=False)
      .str.strip()
      .str.replace(r"\s+"," ",regex=True)
      .str.casefold()
  )
  orders['order_id_key']=clean_key(orders["order_id"])
  returns["order_id_key"]=clean_key(returns["order_id"])
  orders["region_key"]=clean_key(orders["region_key"])
  people["region_key"]=clean_key(people["region_key"])
```

```python
# Merge the Returns Sheet #
returns_for_merge= returns[['order_id','returned']].copy()
print("Duplicate return keys:",returns_for_merge['order_id'].duplicated().sum())
```

```python
# Check Returns has no repeated order #
assert returns_for_merge['order_id'].is_unique
rows_before_merge= len(orders)
orders= orders.merge(
    returns_for_merge,
    on='order_id',
    how='left',
    validate='many_to_one',
    indicator='_return_match'
)
orders['is_returned']=(orders['_return_match']=='both')
orders= orders.drop(columns=['_return_match'])
print("Rows before merge:", rows_before_merge)
print("Rows after merge:", len(orders))
```

```python
# Merge the People Table #
people_for_merge= people[['region','person']].copy()
assert people_for_merge['region'].is_unique
orders= orders.merge(
    people_for_merge,
    on='region',
    how='left',
    validate='many_to_one',
    indicator='_person_match'
)
print(orders['_person_match'].value_counts())
```

```python
# Investigate left-only columns #
orders.loc[orders['_person_match']=='left_only','region'].value_counts()
```

```python
# Investigate Unmatched Records Remove Temporary join Keys #
orders= orders.drop(
    columns=['order_id','region','_person_match'],errors='ignore'
    )
```

```python
# Perform quality checks #
print("Final Shape:",orders.shape)
print("Completed duplicated values:",orders.duplicated().sum())
print("Missing Values By columns:")
print(orders.isna().sum()
      .sort_values(ascending=False)
      .head())
print("Data Types:",orders.dtypes)
```

```python
# Multiply or removing valid order lines #
assert len(orders) ==  rows_before_merge,("Unmached row count change")
assert orders['row_id'].is_unique,("Row ID is not unique")
print("Row ID and Row count checks passed")
```

```python
# Create Final quality report #
quality_report=pd.DataFrame({
   "Column":orders.dtypes,
   "Data Type":[
       str(dtype) for dtype in orders.dtypes
   ],
   "Missing Percentage":[
      round(orders[column].isna().mean()*100,2)
      for column in orders.columns
   ],
  "Unique Values":[
      orders[column].nunique(dropna=True)
      for column in orders.columns
  ]
})
quality_report.to_string(index=False)
quality_report
```

```python
# Missing Column Order_ID Check original file #
import pandas as pd
original_df=pd.read_excel(
    '/content/superstore_2016.xlsx',
    sheet_name='Orders'
)
print(original_df.columns.to_list())
```

```python
# Standardize Columns #
original_df.columns=(
    original_df.columns.str.strip()
    .str.lower()
    .str.replace(r"[^a-z,0-9]+","_",regex=True)
    .str.strip("_")
)
print(original_df.columns.to_list())
```

```python
# Merge the order_id into Cleaned Dataset #
order_lookup= original_df[['row_id','order_id']].copy()
df=df.merge(
    order_lookup,
    on='row_id',
    how='left',
    validate='one_to_one'
)
print(df[['row_id','order_id']].head())
```

```python
# Check the Orders Cleaned Dataset #
print("Order ID column exists:", "order_id" in df.columns)
print("Missing Order IDs:", df["order_id"].isna().sum())
print("Total unique orders:", df["order_id"].nunique())
```

```python
# Calculate Business KPIs #
total_sales = df["sales"].sum()
total_profit = df["profit"].sum()
total_orders = df["order_id"].nunique()
total_customers = df["customer_id"].nunique()
total_products = df["product_name"].nunique()

print("Total Sales:", total_sales)
print("Total Profit:", total_profit)
print("Total Orders:", total_orders)
print("Total Customers:", total_customers)
print("Total Products:", total_products)
```

```python
# Check row_id uniqueness before merging
print("Original duplicate row IDs:",
      original_df["row_id"].duplicated().sum())
region_lookup = original_df[["row_id", "region"]].copy()
df = df.merge(
    region_lookup,
    on="row_id",
    how="left",
    validate="one_to_one"
)
print(df[["row_id", "region"]].head())
print("Missing regions:", df["region"].isna().sum())
```

```python
# Business Scenario 1 #
# 1.Which product category generates the highest sales and profit? #

category_analysis=(
    df.groupby('category')
    .agg(
        Total_Sales=('sales','sum'),
        Total_Profit=('profit','sum')
    )
    .sort_values("Total_Sales",ascending=False)
    )
display(category_analysis)

category_analysis.plot(
    kind='bar',
    figsize=(10,6)
)
plt.title('Sales and Profit by Category')
plt.xlabel('Product Category')
plt.ylabel('Amount')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

Insight: The ______ category generated the highest sales, while the ______ category generated the highest profit. This indicates that revenue and profitability should be evaluated separately.

```python
# Business Scenario-2 #
# 2.Which regions perform well and which need attention? #

region_analysis=(
    df.groupby('region')
    .agg(
        Total_Sales=('sales','sum'),
        Total_Profit=('profit','sum'),
        Unique_Orders=('order_id','nunique')
    )
    .sort_values('Total_Profit',ascending=False)
)
display(region_analysis)

region_analysis.plot(
    kind='bar',
    figsize=(10,6)
)
plt.axvline(0,linestyle='--')
plt.title('Total Profit by Region')
plt.xlabel('Total Profit')
plt.ylabel('Region')
plt.tight_layout()
plt.show()
```

Insight:

The ______ region generated the highest total profit, whereas the ______ region had the lowest. The company should investigate the weaker region's product mix, discounting and operating costs before making decisions

```python
# Business Scenario-3 #
# 3.Are discounts associated with lower profit? #

plt.figure(figsize=(10,6))
sns.scatterplot(
    data=df,
    x='discount',
    y='profit',
    alpha=0.4
)
plt.axhline(0,linestyle='--')
plt.title('Discounts tends to lower Profit')
plt.xlabel('Discount(proportion)')
plt.ylabel('Profit')
plt.tight_layout()
plt.show()
```

```python
# Business Scenario- 4 #
# 4.Compare average profit at different discount levels. #

df['discount_band']=pd.cut(
    df['discount'],
    bins=[-0.001,0.10,0.20,0.30,0.40,1.0],
    labels=[
        '0%',
        'Above 0-10%',
        'Above 10-20%',
        'Above 20-30%',
        'Above 40-50%'
    ]
)
discount_analysis=(
    df.groupby('discount_band',observed=False)
    .agg(
        Order_Lines=('profit','sum'),
        Total_Sales=('sales','sum'),
        Total_Profit=('profit','sum'),
        Average_Profit=('profit','mean')
    )
)
display(discount_analysis)

discount_analysis['Average_Profit'].plot(
    kind='bar',
    color='skyblue',
    figsize=(10,6)
)
plt.axhline(0,linestyle='--',color='red')
plt.title('Average Profit at Different Discount Levels')
plt.xlabel('Discount Band')
plt.ylabel('Average Profit')
plt.tight_layout()
plt.show()
```

```python
# Sns Box Plot #

df['discount_band']=pd.cut(
    df['discount'],
    bins=[-0.001,0.10,0.20,0.30,0.40,1.0],
    labels=[
        '0%',
        'Above 0-10%',
        'Above 10-20%',
        'Above 20-30%',
        'Above 40-50%'
    ]
)
discount_analysis=(
    df.groupby('discount_band',observed=False)
    .agg(
        Order_Lines=('profit','size'),
        Total_Sales=('sales','sum'),
        Total_Profit=('profit','sum'),
        Average_Profit=('profit','mean')
    )
)
display(discount_analysis)
plt.figure(figsize=(10,6))
sns.boxplot(
    data=df,
    x='discount_band',
    y='profit',
    showfliers=False,
    color='skyblue'
)
plt.axhline(0,linestyle='--',color='red')
plt.title('Profit Distribution at Discount Levels')
plt.xlabel('Discount Band')
plt.ylabel('Average Profit')
plt.legend()
plt.xticks(rotation=20)
plt.tight_layout()
plt.show()
```

```python
# Business Scenario-5 #
# 5. Which months generate the most sales? Is there a seasonal trend or a period of declining sales? #

df['order_date']= pd.to_datetime(df['order_date'],errors='coerce')
df['year_month']= df['order_date'].dt.to_period('M')
monthly_sales=(
    df.dropna(subset=['order_date'])
    .groupby('year_month')['sales']
    .sum()
)
plt.figure(figsize=(14,6))
plt.plot(
    monthly_sales.index.astype(str),
    monthly_sales.values,
    color='red',
    marker='o'
)
plt.title('Monthly Sales Trend')
plt.xlabel('Month')
plt.ylabel('Total Sales')
plt.xticks(rotation=45)
plt.grid(alpha=0.3)
plt.legend()
plt.tight_layout()
plt.show()

print("Highest-sales month:")
print(monthly_sales.idxmax(), monthly_sales.max())

print("\nLowest-sales month:")
print(monthly_sales.idxmin(), monthly_sales.min())
```

```python
# Business Scenario-6 #
# Which products have the lowest profit,and are they also high-sales products?#

product_analysis=(
    df.groupby('product_name')
    .agg(
        Total_Sales=('sales','sum'),
        Total_Profit=('profit','sum'),
        Total_Quantity=('quantity','sum')
    )
)
lowest_profit=product_analysis.sort_values(
    'Total_Profit',
    ascending=True
).head(10)
display(lowest_profit)

plt.figure(figsize=(12,7))
plt.barh(
    lowest_profit.index,
    lowest_profit['Total_Profit'],
    color='skyblue'
)
plt.axvline(0,linestyle='--')
plt.title('Products with lowest total profit')
plt.xlabel('Sales')
plt.ylabel('Total Profit')
plt.tight_layout()
plt.show()
```

```python
# Business Scenario-7 #
# Which products sell well but have poor profitability? #

product_analysis=(
    df.groupby('product_name')
    .agg(
        Total_Sales=('sales','sum'),
        Total_Profit=('profit','sum'),
        Total_Quantity=('quantity','sum')
    )
)
sns.scatterplot(
    data=product_analysis.reset_index(),
    x='Total_Sales',
    y='Total_Profit',
    alpha=0.7
)
plt.axhline(0,linestyle='--',color='red')
plt.title('Products Sales Vs Total_Profit')
plt.xlabel('Total Sales')
plt.ylabel('Total Profit')
plt.grid()
plt.tight_layout()
plt.show()

print('Lowest Profitability:',product_analysis.min())
```

```python
# Business Scenario-8 #
# How many unique orders were returned, and what percentage of all orders were returned? #

return_summary=(
    df.groupby('order_id')['is_returned']
    .max()
)
total_orders= return_summary.size
returned_orders= int(return_summary.sum())
return_rate=(
    returned_orders/total_orders*100
    if total_orders else 0
)

not_returned_orders=(total_orders - returned_orders)
labels=['Returned Orders','Non-Returned Orders']
values=[returned_orders,not_returned_orders]
plt.pie(
    values,
    labels=labels,
    autopct="%1.1f%%",
    startangle=90,
    wedgeprops={'width':0.4}
)
plt.title('Returned Orders Vs Non-Returned Orders')
plt.tight_layout()
plt.show()

print('Total Orders:',total_orders)
print("Returned Orders:",returned_orders)
print(f"Return rate Percentage:{return_rate:.2f}%")
```

```python
# Business Scenario-9 #
# How many days does an order take to ship, and does delivery time differ by shipping mode? #

df['order_date']=pd.to_datetime(df['order_date'],errors='coerce')
df['ship_date']=pd.to_datetime(df['ship_date'],errors='coerce')
df['shipping_days']= (df['ship_date'] - df['order_date']).dt.days

valid_shipping=df[
    df['shipping_days'].notna() & (df['shipping_days']>=0)].copy()
shipping_analysis=(
    valid_shipping.groupby('ship_mode')
    .agg(
        Order_Count=('order_id','nunique'),
        Average_Shipping_Days=('shipping_days','mean'),
        Median_Shipping_Days=('shipping_days','median')
    )
    .sort_values('Average_Shipping_Days')
)
display(shipping_analysis.round(2))

sns.boxplot(
    data=df,
    x='ship_mode',
    y='shipping_days'
)
plt.title('Shipping Duration by Shipping Mode')
plt.xlabel('Shipping Mode')
plt.ylabel('Days Between Order and Shipment')
plt.xticks(rotation=15)
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from matplotlib.ticker import FuncFormatter

# 1. CHECK REQUIRED COLUMNS

required_columns = [
    "sales", "profit", "order_id", "customer_id",
    "product_name", "order_date", "region",
    "category", "discount", "is_returned"
]

missing_columns = [col for col in required_columns if col not in df.columns]

if missing_columns:
    raise ValueError(
        f"Missing columns: {missing_columns}. "
        "Restore them from the original Orders sheet "
        "or your Returns data before creating the dashboard."
    )

# Work with a copy so the original df is not changed
data = df.copy()

data["order_date"] = pd.to_datetime(
    data["order_date"], errors="coerce"
)


# 2. CALCULATE KPIs

kpis = {
    "Total Sales": data["sales"].sum(),
    "Total Profit": data["profit"].sum(),
    "Total Orders": data["order_id"].nunique(),
    "Customers": data["customer_id"].nunique()
}


# 3. PREPARE CHART DATA


# Monthly sales trend
monthly_sales = (
    data.dropna(subset=["order_date"])
    .assign(Month=lambda x: x["order_date"].dt.to_period("M"))
    .groupby("Month")["sales"]
    .sum()
)

# Profit by region
region_profit = (
    data.groupby("region")["profit"]
    .sum()
    .sort_values()
)

# Sales by category
category_sales = (
    data.groupby("category")["sales"]
    .sum()
    .sort_values(ascending=False)
)

# Product sales vs profit: top 30 products by sales
product_summary = (
    data.groupby("product_name")
    .agg(
        Sales=("sales", "sum"),
        Profit=("profit", "sum")
    )
    .nlargest(30, "Sales")
)

# Discount bands
data["discount_band"] = pd.cut(
    data["discount"],
    bins=[-0.001, 0, 0.10, 0.20, 0.30, 0.40, 1.0],
    labels=[
        "0%", "0–10%", "10–20%",
        "20–30%", "30–40%", "40%+"
    ],
    include_lowest=True
)

discount_profit = (
    data.groupby("discount_band", observed=False)["profit"]
    .mean()
)

# Return rate calculated at order level
order_returns = (
    data.groupby("order_id")["is_returned"]
    .max()
    .fillna(False)
    .astype(bool)
)

returned_count = int(order_returns.sum())
not_returned_count = len(order_returns) - returned_count

return_rate = (
    returned_count / len(order_returns) * 100
    if len(order_returns) > 0 else 0
)


# 4. BUILD ONE DASHBOARD FIGURE

sns.set_theme(style="whitegrid")

fig = plt.figure(figsize=(20, 17))

grid = fig.add_gridspec(
    4, 2,
    height_ratios=[0.65, 1.4, 1.4, 1.4],
    hspace=0.42,
    wspace=0.25
)

fig.suptitle(
    "GLOBAL SUPERSTORE — BUSINESS KPI DASHBOARD",
    fontsize=22,
    fontweight="bold",
    y=0.99
)

# 5. KPI CARDS AT THE TOP

kpi_grid = grid[0, :].subgridspec(1, 4, wspace=0.15)

for i, (metric, value) in enumerate(kpis.items()):
    ax = fig.add_subplot(kpi_grid[0, i])
    ax.set_facecolor("#EAF2F8")

    if metric in ["Total Sales", "Total Profit"]:
        display_value = f"${value:,.0f}"
    else:
        display_value = f"{value:,.0f}"

    ax.text(
        0.5, 0.65, display_value,
        ha="center", va="center",
        fontsize=22, fontweight="bold",
        transform=ax.transAxes
    )

    ax.text(
        0.5, 0.25, metric,
        ha="center", va="center",
        fontsize=13,
        transform=ax.transAxes
    )

    ax.set_xticks([])
    ax.set_yticks([])

    for spine in ax.spines.values():
        spine.set_visible(False)

# 6. CHART 1: MONTHLY SALES TREND

ax1 = fig.add_subplot(grid[1, 0])

ax1.plot(
    monthly_sales.index.astype(str),
    monthly_sales.values,
    marker="o",
    linewidth=2
)

ax1.set_title("Monthly Sales Trend", fontweight="bold")
ax1.set_xlabel("Month")
ax1.set_ylabel("Sales")
ax1.yaxis.set_major_formatter(
    FuncFormatter(lambda x, _: f"${x/1000:.0f}K")
)
ax1.tick_params(axis="x", rotation=60)


# 7. CHART 2: PROFIT BY REGION

ax2 = fig.add_subplot(grid[1, 1])

region_profit.plot(
    kind="barh",
    ax=ax2,
    color=[
        "#C0392B" if x < 0 else "#3498DB"
        for x in region_profit.values
    ]
)

ax2.axvline(0, color="black", linewidth=1)
ax2.set_title("Profit by Region", fontweight="bold")
ax2.set_xlabel("Total Profit")
ax2.set_ylabel("Region")

# 8. CHART 3: SALES BY CATEGORY

ax3 = fig.add_subplot(grid[2, 0])

category_sales.plot(
    kind="bar",
    ax=ax3,
    color="#48C9B0"
)

ax3.set_title("Sales by Category", fontweight="bold")
ax3.set_xlabel("Category")
ax3.set_ylabel("Total Sales")
ax3.yaxis.set_major_formatter(
    FuncFormatter(lambda x, _: f"${x/1000:.0f}K")
)
ax3.tick_params(axis="x", rotation=15)


# 9. CHART 4: PRODUCT SALES VS PROFIT

ax4 = fig.add_subplot(grid[2, 1])

sns.scatterplot(
    data=product_summary,
    x="Sales",
    y="Profit",
    ax=ax4,
    s=70
)

ax4.axhline(0, color="red", linestyle="--")
ax4.set_title(
    "Product Sales vs Profit (Top 30 by Sales)",
    fontweight="bold"
)
ax4.set_xlabel("Total Sales")
ax4.set_ylabel("Total Profit")


# 10. CHART 5: PROFIT BY DISCOUNT BAND

ax5 = fig.add_subplot(grid[3, 0])

discount_profit.plot(
    kind="bar",
    ax=ax5,
    color="#AF7AC5"
)

ax5.axhline(0, color="red", linestyle="--")
ax5.set_title("Average Profit by Discount Band", fontweight="bold")
ax5.set_xlabel("Discount Band")
ax5.set_ylabel("Average Profit")
ax5.tick_params(axis="x", rotation=20)

# 11. CHART 6: RETURNED-ORDER RATE

ax6 = fig.add_subplot(grid[3, 1])

ax6.pie(
    [returned_count, not_returned_count],
    labels=["Returned", "Not Returned"],
    autopct="%1.1f%%",
    startangle=90,
    wedgeprops={"width": 0.42}
)

ax6.set_title(
    f"Returned-Order Rate: {return_rate:.2f}%",
    fontweight="bold"
)

# 12. DISPLAY THE COMPLETE DASHBOARD

plt.show()
```
